```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/final_scored_dataset.csv")

latest = (
    df.sort_values("year")
      .groupby("district")
      .tail(1)
)

latest = latest.sort_values(
    "urban_pulse_risk_score",
    ascending=False
)

plt.figure(figsize=(10,6))

sns.barplot(
    data=latest,
    x="district",
    y="urban_pulse_risk_score"
)

plt.title("Urban Pulse Risk Score by District")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.lineplot(
    data=df,
    x="year",
    y="avg_aqi",
    hue="district",
    marker="o"
)

plt.title("AQI Trend by District")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.lineplot(
    data=df,
    x="year",
    y="urban_pulse_risk_score",
    hue="district",
    marker="o"
)

plt.title("Urban Pulse Risk Score Trend")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.regplot(
    data=df,
    x="density",
    y="avg_aqi",
    scatter=False,
    color="red"
)

plt.title("AQI vs Population Density")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[
        [
            "avg_aqi",
            "population",
            "density",
            "literacy",
            "healthcare_facilities",
            "urban_pulse_risk_score"
        ]
    ].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.regplot(
    data=df,
    x="healthcare_facilities",
    y="avg_aqi",
    scatter=False,
    color="red"
)

plt.title("AQI vs Healthcare Facilities")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="risk_category",
    order=["Low","Medium","High"]
)

plt.title("Distribution of Risk Categories")

plt.tight_layout()
plt.show()
```
