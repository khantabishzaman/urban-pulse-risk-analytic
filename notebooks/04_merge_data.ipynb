```python
import pandas as pd
```

```python
aqi = pd.read_csv("../data/processed/district_aqi.csv")
census = pd.read_csv("../data/processed/district_census.csv")
healthcare = pd.read_csv("../data/processed/district_healthcare.csv")

final_df = (
    aqi
    .merge(census, on="district", how="left")
    .merge(healthcare, on="district", how="left")
)

final_df.head()
```

```python
final_df.isnull().sum()
```

```python
final_df.to_csv(
    "../data/processed/final_dataset.csv",
    index=False
)

print("final_dataset.csv saved successfully")
```

```python
final_df["avg_aqi"] = final_df["avg_aqi"].round(2)
```
