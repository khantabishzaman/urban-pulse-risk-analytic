# Urban Pulse Risk Assessment: Integrating Air Quality, Demographic, and Healthcare Indicators Across Delhi Districts

## 1. Project Overview

Rapid urbanization has significantly impacted environmental quality and public health in metropolitan cities. Air pollution, population density, literacy levels, and healthcare infrastructure collectively influence the quality of life and vulnerability of urban populations.

This project introduces an **Urban Pulse Risk Score**, a composite indicator developed by integrating Air Quality Index (AQI), demographic characteristics, and healthcare infrastructure across selected districts of Delhi. The objective is to identify vulnerable districts and support data-driven urban planning and policy decisions.

---

# 2. Objectives

The primary objectives of this project are:

1. To analyze AQI trends across major monitoring stations in Delhi.
2. To map AQI stations to their respective districts.
3. To integrate environmental, demographic, and healthcare datasets.
4. To develop a composite Urban Pulse Risk Score.
5. To identify high-risk and low-risk districts.
6. To provide actionable recommendations for policymakers and urban planners.

---

# 3. Data Sources

## 3.1 Air Quality Data

Air Quality Index (AQI) data was collected from five major monitoring stations:

* Anand Vihar
* Ashok Vihar
* Chandni Chowk
* Mundka
* Najafgarh

The data spans multiple years between 2017 and 2023 and was aggregated into annual district-level AQI values.

---

## 3.2 Census Data

District-level demographic indicators were obtained from Census and Planning Department records.

Variables included:

* Population
* Population Density
* Literacy Rate

---

## 3.3 Healthcare Infrastructure Data

Healthcare facility counts were extracted from district-level healthcare records and government health directories.

Variables included:

* Number of healthcare facilities per district

## 4.2 Urban Pulse Risk Score

The Urban Pulse Risk Score was computed using normalized indicators.

### Risk-Contributing Factors

* AQI
* Population Density

### Risk-Reducing Factors

* Literacy Rate
* Healthcare Facilities

Weights assigned:

| Indicator             | Weight |
| --------------------- | ------ |
| AQI                   | 40%    |
| Density               | 30%    |
| Healthcare Facilities | 20%    |
| Literacy Rate         | 10%    |

The resulting score ranges from 0 to 100, where higher values indicate greater urban vulnerability.

# 6. Key Findings

## Finding 1: East District Exhibits the Highest Urban Risk

The Urban Pulse Risk Score identified East District as the most vulnerable district among the analyzed regions.

Risk Score Ranking:

1. East
2. Central
3. West
4. North West
5. South West

East District consistently recorded the highest AQI values and one of the highest population densities, significantly increasing its overall risk score.

---

## Finding 2: Air Quality Has Improved Over Time

AQI trends show a gradual decline in pollution levels across most districts between 2017 and 2023.

Examples:

* East District AQI decreased from approximately 376 to 263.
* Central District AQI decreased from approximately 338 to 196.
* North West and South West also showed improvements.

This suggests that pollution-control measures may be contributing positively to air quality improvement.

---

## Finding 3: Population Density is the Strongest Risk Driver

Correlation analysis revealed:

Density ↔ AQI = 0.68

Density ↔ Urban Pulse Risk Score = 0.94

This indicates that highly dense districts tend to experience greater environmental stress and higher overall risk.

---

## Finding 4: Healthcare Infrastructure Reduces Vulnerability

Healthcare Facilities ↔ AQI = -0.37

Healthcare Facilities ↔ Urban Pulse Risk Score = -0.63

Districts with greater healthcare capacity generally displayed lower risk levels.

Healthcare infrastructure acts as an important mitigating factor against urban vulnerability.

---

## Finding 5: AQI is the Dominant Component of Risk

AQI ↔ Urban Pulse Risk Score = 0.85

This confirms that air quality remains the most influential factor in determining urban risk within the study area.

# 9. Limitations

The study has several limitations:

1. Only five AQI monitoring stations were analyzed.
2. Census variables remain static over the study period.
3. Healthcare data was available only as facility counts.
4. Additional variables such as income, traffic density, and industrial activity were not included.

Future studies can incorporate these variables to improve predictive accuracy.

---

# 10. Conclusion

This project successfully developed an Urban Pulse Risk Score by integrating environmental, demographic, and healthcare indicators across selected districts of Delhi.

The analysis revealed that East District consistently exhibits the highest urban risk due to elevated AQI levels and high population density. Population density emerged as the strongest contributing factor to vulnerability, while healthcare infrastructure demonstrated a mitigating effect.

The Urban Pulse Risk Score provides a simple, interpretable, and scalable framework for identifying vulnerable districts and supporting evidence-based urban planning decisions. The methodology can be extended to additional districts, cities, and indicators to support sustainable urban development initiatives.
